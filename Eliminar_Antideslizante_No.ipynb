```python
df2['ANTIDESLIZANTE_NO'] = df2["Descripción"].str.contains('Antideslizante NO')
df2["Descripción"] = df2["Descripción"].str.replace(
    r'.*Antideslizante NO.*(\n|$)',
    '',
    regex=True
)
df2.head()
```

```python
df3 = df2[df2.ANTIDESLIZANTE_NO.fillna(False)]
df3["Descripción"] = df3["Descripción"].str.replace("\n", "<br>", regex=False)
```

```python
df1 = pd.read_excel("https://drive.usercontent.google.com/download?id=", index_col=0)
```

```python
df1.head()
```

```python
df2 = df1.drop(['Short Description'], axis=1)
df2.head()
```

```python
# Modifica la columna 'Content' dentro del DataFrame df2, manteniendo otras columnas
df2['Content'] = df2['Content'].str.replace('<br>', '_').str.replace('<h3>', '_').str.replace('<b>', '_')
df2.head()
```

```python
df2.index.str.contains("Porc.")
df2.head()
```

```python
df2['PORC'] = df2.index.str.contains('Porc\\.')
df2.head()
```

```python
import re

# Asegura de que la columna 'Content' sea string antes de limpiar y aplica la limpieza
df2['Content'] = df2['Content'].astype(str).apply(lambda x: re.sub(r'<.*?>', '', x))

# Guardar el DataFrame df2 (con la columna 'Content' limpia y todas las demás columnas) a Excel
df2.to_excel('PORC.xlsx', index=True)
```
